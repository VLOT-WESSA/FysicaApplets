<div>    <font color=#690027 markdown="1">        <h2>1. NumPy-arrays</h2>     </font></div>

<div class="alert alert-block alert-info">Een <b>NumPy-array</b> is een rij of tabel van getallen van hetzelfde type. Een gewone rij is een <b>1D-array</b>; een tabel (matrix) is een <b>2D-array</b>. Je maakt ze met <code>np.array(...)</code>. Elk array-object heeft kenmerken:<ul><li><code>ndim</code>: het aantal dimensies (1 of 2)</li><li><code>shape</code>: (aantal rijen, aantal kolommen)</li><li><code>size</code>: het totale aantal elementen</li><li><code>dtype</code>: het type van de elementen</li></ul></div>

### Voorbeeld 1.1 &mdash; Een matrix (2D-array)De matrix $\begin{bmatrix} 1 & 2 & 0 \\ 3 & 4 & 5 \end{bmatrix}$ heeft 2 rijen en 3 kolommen.

```python
matrix = np.array([[1, 2, 0], [3, 4, 5]])   # 2 rijen, 3 kolommen

print(matrix.ndim)     # 2  (tweedimensionaal)
print(matrix.shape)    # (2, 3)
print(matrix.size)     # 6
print(matrix.dtype)    # int64
```

### Voorbeeld 1.2 &mdash; Een rij (1D-array) en rekenenOp een array reken je *elementgewijs*: elke bewerking gebeurt op alle elementen tegelijk.

```python
rij = np.array([1, 2, 3, 4, 5, 6])

print(rij * 2)        # elk element maal 2
print(rij + 10)       # bij elk element 10 optellen
print(np.sum(rij))    # som
print(np.mean(rij))   # gemiddelde
```

<div class="alert alert-block alert-info">Handige functies om een reeks getallen te maken:<br>&bull; <code>np.arange(begin, einde, stap)</code>: zoals <code>range</code>, maar geeft een array (mag kommagetallen);<br>&bull; <code>np.linspace(begin, einde, aantal)</code>: <i>aantal</i> gelijk verdeelde waarden tussen begin en einde.</div>

### Voorbeeld 1.3 &mdash; Reeksen maken

```python
print(np.arange(0, 2, 0.5))     # [0.  0.5 1.  1.5]
print(np.linspace(0, 10, 5))    # [ 0.   2.5  5.   7.5 10. ]
```

<div>    <font color=#690027 markdown="1">        <h2>2. Het tekenvenster</h2>     </font></div>

<div class="alert alert-block alert-info">Voor grafieken gebruik je <code>matplotlib.pyplot</code>, ingevoerd als <code>plt</code>. Je maakt eerst een <b>tekenvenster</b> met <code>plt.figure()</code>, tekent dan een grafiek, en toont alles met <code>plt.show()</code>. Voeg altijd een titel en asnamen toe.</div>

### De nodige module importeren

```python
import matplotlib.pyplot as plt
```

### Voorbeeld 2.1 &mdash; Een functie plotten
We tekenen $y = x^2$ voor x van $-5$ tot $5$.

```python
x = np.linspace(-5, 5, 100)     # 100 x-waarden
y = x ** 2                      # bijhorende y-waarden

plt.figure(figsize=(6, 4))      # tekenvenster van 6 op 4 inch
plt.plot(x, y)                  # lijngrafiek
plt.title("De parabool y = x^2")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

<div class="alert alert-block alert-warning">Zo teken je elke wiskundige of natuurkundige functie: een reeks x-waarden met <code>linspace</code>, de bijhorende y-waarden via de formule, en <code>plot</code>. Handig om bijvoorbeeld een baan, een golf of een vervalcurve zichtbaar te maken.</div>
