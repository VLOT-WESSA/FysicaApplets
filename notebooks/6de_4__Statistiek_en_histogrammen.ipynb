### De nodige modules importeren

```python
import numpy as np
import matplotlib.pyplot as plt
```

<div>    <font color=#690027 markdown="1">        <h2>1. Centrummaten: gemiddelde en mediaan</h2>     </font></div>

<div class="alert alert-block alert-info">Het <b>gemiddelde</b> (<code>np.mean</code>) is de som gedeeld door het aantal. De <b>mediaan</b> (<code>np.median</code>) is de middelste waarde na sorteren. De mediaan is minder gevoelig voor <b>uitschieters</b>.</div>

### Voorbeeld 1.1 &mdash; Een meetreeks samenvatten

```python
metingen = np.array([20.1, 20.4, 19.8, 20.0, 20.3, 45.0])   # let op de uitschieter 45.0

print("gemiddelde:", np.mean(metingen))
print("mediaan:   ", np.median(metingen))
```

<div class="alert alert-block alert-warning">Eén foute meting (een <b>uitschieter</b>) trekt het gemiddelde sterk omhoog, maar laat de mediaan bijna ongemoeid. Daarom rapporteren wetenschappers vaak beide, of verwijderen ze eerst duidelijke meetfouten.</div>

<div>    <font color=#690027 markdown="1">        <h2>2. Spreidingsmaten</h2>     </font></div>

<div class="alert alert-block alert-info">De <b>standaardafwijking</b> (<code>np.std</code>) meet hoe ver de metingen gemiddeld van het gemiddelde liggen: klein betekent nauwkeurig, groot betekent veel spreiding. Het <b>bereik</b> is het verschil tussen de grootste en de kleinste waarde.</div>

### Voorbeeld 2.1 &mdash; Spreiding berekenen

```python
metingen = np.array([20.1, 20.4, 19.8, 20.0, 20.3, 19.9])

print("gemiddelde:        ", round(np.mean(metingen), 3))
print("standaardafwijking:", round(np.std(metingen), 3))
print("bereik:            ", np.max(metingen) - np.min(metingen))
```

<div>    <font color=#690027 markdown="1">        <h2>3. Het histogram</h2>     </font></div>

<div class="alert alert-block alert-info">Een <b>histogram</b> verdeelt de metingen in <b>klassen</b> (bins) en toont hoe vaak een waarde in elke klasse voorkomt. Je tekent het met <code>plt.hist(data, bins=aantal)</code>. Zo zie je de vorm van de verdeling.</div>

### Voorbeeld 3.1 &mdash; Een histogram van meetgegevens

```python
metingen = np.array([21, 22, 22, 23, 23, 23, 24, 24, 24, 24, 25, 25, 25, 26, 26, 27])

plt.figure(figsize=(6, 4))
plt.hist(metingen, bins=7, edgecolor="black")
plt.title("Verdeling van de metingen")
plt.xlabel("waarde")
plt.ylabel("frequentie")
plt.show()
```

<div>    <font color=#690027 markdown="1">        <h2>4. De normale verdeling</h2>     </font></div>

<div class="alert alert-block alert-info">Veel meetgegevens volgen bij benadering een <b>normale verdeling</b> (klokvorm): de meeste waarden liggen rond het gemiddelde, met symmetrisch afnemende frequenties. Je kan zo'n verdeling <b>simuleren</b> met <code>np.random.normal(gemiddelde, standaardafwijking, aantal)</code>.</div>

### Voorbeeld 4.1 &mdash; Meetfouten simuleren
We bootsen 1000 metingen na van een lengte van 100 mm met een meetonzekerheid (standaardafwijking) van 2 mm.

```python
metingen = np.random.normal(100, 2, 1000)   # gemiddelde 100, spreiding 2, 1000 metingen

print("gemeten gemiddelde:", round(np.mean(metingen), 2))
print("gemeten spreiding: ", round(np.std(metingen), 2))

plt.figure(figsize=(6, 4))
plt.hist(metingen, bins=30, edgecolor="black")
plt.title("Genormeerde meetreeks (klokvorm)")
plt.xlabel("gemeten lengte (mm)")
plt.ylabel("frequentie")
plt.show()
```

<div class="alert alert-block alert-warning">Herhaal je een meting vaak, dan spreiden de resultaten zich rond de echte waarde volgens een <b>klokcurve</b>. Het gemiddelde benadert de echte waarde; de standaardafwijking beschrijft de meetonzekerheid.</div>
