<div>    <font color=#690027 markdown="1">        <h2>1. Wat is een Jupyter Notebook?</h2>     </font></div>

<div class="alert alert-block alert-info">Een <b>Jupyter Notebook</b> is een document dat bestaat uit <b>cellen</b>. Er zijn twee soorten cellen:<br>&bull; een <b>code-cel</b>: hierin schrijf je Python-code die de computer uitvoert;<br>&bull; een <b>Markdown-cel</b> (tekstcel): hierin staat uitleg, opgemaakt met tekst, titels en formules.<br><br>Je herkent een code-cel aan het label <code>[ ]</code> links ervan. Een notebook heeft de extensie <code>.ipynb</code>.</div>

<div>    <font color=#690027 markdown="1">        <h2>2. De gekleurde kaders</h2>     </font></div>

<div class="alert alert-block alert-info">In alle notebooks gebruiken we vier kleuren kaders. Zo weet je meteen wat voor soort informatie je leest.</div>

<div class="alert alert-box alert-success">Een <b>groen</b> kader geeft de inleiding, de leerdoelen of een samenvatting van de notebook.</div>

<div class="alert alert-block alert-info">Een <b>blauw</b> kader bevat de <b>theorie</b>: de begrippen en regels die je moet kennen.</div>

<div class="alert alert-block alert-warning">Een <b>geel</b> kader geeft extra achtergrond of een toepassing, vaak uit de wetenschappen.</div>

<div class="alert alert-block alert-danger">Een <b>rood</b> kader waarschuwt voor een veelgemaakte fout of geeft een belangrijke tip.</div>

<div>    <font color=#690027 markdown="1">        <h2>3. Cellen uitvoeren en bewerken</h2>     </font></div>

<div class="alert alert-block alert-info"><b>Code uitvoeren</b><br>&bull; <code>Ctrl + Enter</code>: voer de cel uit en blijf erin staan.<br>&bull; <code>Shift + Enter</code>: voer de cel uit en ga naar de volgende cel.<br><br><b>Cellen bewerken</b> (klik eerst links naast een cel, zodat ze blauw omrand is):<br>&bull; <code>A</code>: nieuwe cel <b>boven</b> &nbsp;&bull; <code>B</code>: nieuwe cel <b>onder</b><br>&bull; <code>D D</code> (twee keer): cel <b>verwijderen</b><br>&bull; <code>M</code>: cel wordt <b>Markdown</b> (tekst) &nbsp;&bull; <code>Y</code>: cel wordt <b>code</b><br>&bull; <code>Z</code>: laatste bewerking ongedaan maken.</div>
