<div>    <font color=#690027 markdown="1">        <h2>2. Een experiment simuleren</h2>     </font></div>

<div class="alert alert-block alert-info">Door een experiment <b>heel vaak</b> te herhalen, benadert de <b>relatieve frequentie</b> (aantal keer gebeurd / totaal aantal) de echte <b>kans</b>. Dit heet de <i>wet van de grote getallen</i>. Met <code>np.sum(voorwaarde)</code> tel je hoe vaak een voorwaarde waar is.</div>

### Voorbeeld 2.1 &mdash; Kans op een zes
We werpen 100 000 keer en tellen de zessen.

```python
worpen = np.random.randint(1, 7, 100000)
aantal_zes = np.sum(worpen == 6)          # telt de zessen
print("relatieve frequentie:", aantal_zes / 100000)
print("theoretische kans:   ", round(1/6, 5))
```

<div class="alert alert-block alert-warning">Hoe vaker je simuleert, hoe dichter de relatieve frequentie bij de theoretische kans $\tfrac16$ komt. Simulaties laten je zo kansen schatten die je niet makkelijk met de hand berekent.</div>

<div>    <font color=#690027 markdown="1">        <h2>3. Radioactief verval simuleren</h2>     </font></div>

<div class="alert alert-block alert-info">Verval is een <b>toevalsproces</b>: elk atoom heeft per tijdstap dezelfde kans om te vervallen. We simuleren dat voor veel atomen tegelijk: per stap trekken we voor elk niet-vervallen atoom een toevalsgetal en laten het vervallen als dat getal kleiner is dan de vervalkans.</div>

### Voorbeeld 3.1 &mdash; Verval van 10 000 atomen

```python
aantal = 10000            # startaantal atomen
vervalkans = 0.1          # kans per tijdstap dat een atoom vervalt
stappen = 40

overgebleven = []
for stap in range(stappen):
    overgebleven.append(aantal)
    # trek voor elk overgebleven atoom een toevalsgetal; vervalt als < vervalkans
    trekkingen = np.random.random(aantal)
    vervallen = np.sum(trekkingen < vervalkans)
    aantal = aantal - vervallen

plt.figure(figsize=(6, 4))
plt.plot(range(stappen), overgebleven)
plt.title("Radioactief verval (simulatie)")
plt.xlabel("tijdstap")
plt.ylabel("aantal atomen")
plt.show()
```

<div class="alert alert-block alert-warning">De simulatie geeft vanzelf de bekende <b>exponentiële vervalcurve</b>, zonder dat je de formule invoert. Zo laat toeval op microniveau een vaste wet op macroniveau ontstaan &mdash; precies wat er in een radioactief staal gebeurt.</div>

<div>    <font color=#690027 markdown="1">        <h2>4. Het getal $\pi$ schatten</h2>     </font></div>

<div class="alert alert-block alert-info">Een klassieke Monte-Carlotoepassing: gooi willekeurige punten in een vierkant van $1 \times 1$ en tel hoeveel er binnen de kwartcirkel met straal 1 vallen (waar $x^2 + y^2 \le 1$). De verhouding benadert $\tfrac{\pi}{4}$, dus $\pi \approx 4 \cdot \frac{\text{punten in cirkel}}{\text{totaal}}$.</div>

### Voorbeeld 4.1 &mdash; π benaderen met toeval

```python
n = 100000
x = np.random.random(n)
y = np.random.random(n)

in_cirkel = np.sum(x**2 + y**2 <= 1)      # tel de punten binnen de kwartcirkel
schatting = 4 * in_cirkel / n
print("schatting van pi:", schatting)
```
