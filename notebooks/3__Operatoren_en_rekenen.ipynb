<div>    <font color=#690027 markdown="1">        <h2>1. Rekenkundige operatoren</h2>     </font></div>

<div class="alert alert-block alert-info">De gewone wiskundige operatoren bestaan ook in Python:<table><tr><th>bewerking</th><th>wiskunde</th><th>Python</th></tr><tr><td>optellen / aftrekken</td><td>+ , &minus;</td><td><code>+</code> , <code>-</code></td></tr><tr><td>vermenigvuldigen</td><td>&middot;</td><td><code>*</code></td></tr><tr><td>delen</td><td>/</td><td><code>/</code></td></tr><tr><td>gehele deling</td><td></td><td><code>//</code></td></tr><tr><td>rest (modulo)</td><td>mod</td><td><code>%</code></td></tr><tr><td>macht</td><td>$a^{b}$</td><td><code>a ** b</code></td></tr></table>Python respecteert de <b>volgorde van de bewerkingen</b>: eerst machten, dan maal en delen, dan plus en min. Met haakjes forceer je een andere volgorde.</div>

### Voorbeeld 1.1 &mdash; Volgorde van de bewerkingen

```python
print(2 + 7 * 8)      # eerst maal -> 58
print((2 + 7) * 8)    # haakjes eerst -> 72
print(2 ** 5)         # macht -> 32
print(-3 ** 2)        # macht eerst, dan min -> -9
```

<div class="alert alert-block alert-danger"><b>Let op &mdash; macht en minteken.</b> <code>-3 ** 2</code> geeft <code>-9</code>, want de macht wordt eerst berekend en pas daarna komt het minteken. Bedoel je $(-3)^2 = 9$, schrijf dan <code>(-3) ** 2</code>.</div>

<div>    <font color=#690027 markdown="1">        <h2>2. Delen: / , // en %</h2>     </font></div>

<div class="alert alert-block alert-info">Python kent drie soorten deling:<br>&bull; <code>/</code> is de <b>gewone deling</b> en geeft altijd een <b>float</b>;<br>&bull; <code>//</code> is de <b>gehele deling</b>: hoeveel keer past de deler volledig?<br>&bull; <code>%</code> (<b>modulo</b>) geeft de <b>rest</b> van de gehele deling.<br>Zo is <code>17 // 4 = 4</code> en <code>17 % 4 = 1</code>, want $17 = 4\cdot 4 + 1$.</div>

### Voorbeeld 2.1 &mdash; Het verschil tonen

```python
print(17 / 4)     # gewone deling -> 4.25
print(17 // 4)    # gehele deling  -> 4
print(17 % 4)     # rest (modulo)  -> 1
```

<div class="alert alert-block alert-warning">De modulo <code>%</code> is heel handig in de wetenschappen: <code>getal % 2 == 0</code> test of een getal <b>even</b> is, en met modulo bepaal je bijvoorbeeld of een jaartal een schrikkeljaar is of hoeveel volledige porties je uit een hoeveelheid haalt.</div>

<div class="alert alert-block alert-info">Je kan een toekenning <b>verkort</b> schrijven. In plaats van <code>x = x + 3</code> schrijf je <code>x += 3</code>. Dit werkt ook met <code>-= , *= , /=</code> enzovoort.</div>

<div>    <font color=#690027 markdown="1">        <h2>3. De module math</h2>     </font></div>

<div class="alert alert-block alert-info">Voor meer dan de basisbewerkingen gebruik je de <b>module math</b>. Je voert ze in met <code>import math</code> en roept een functie op met <code>math.functie(...)</code>. Enkele nuttige:<ul><li><code>math.sqrt(x)</code>: vierkantswortel</li><li><code>math.pi</code>: het getal $\pi$</li><li><code>math.sin(x)</code>, <code>math.cos(x)</code>, <code>math.tan(x)</code>: hoek in radialen</li><li>ingebouwd (geen import nodig): <code>round(x, n)</code> en <code>abs(x)</code></li></ul></div>

### Voorbeeld 3.1 &mdash; De inhoud van een bol
De inhoud van een bol met straal $r$ is $V = \tfrac{4}{3}\pi r^{3}$.

```python
import math

straal = 7                                  # in cm
inhoud = 4/3 * math.pi * straal ** 3
print("inhoud:", round(inhoud, 2), "cm3")   # afgerond op 2 decimalen
```

<div class="alert alert-block alert-warning">Veel natuurwetten zijn formules die je zo in Python overtypt: de inhoud van een bol, de omtrek van een cirkel, de slingertijd $T = 2\pi\sqrt{\ell/g}$... Met de module math heb je $\pi$ en de wortel meteen bij de hand.</div>

<div>    <font color=#690027 markdown="1">        <h2>4. Vergelijkingsoperatoren</h2>     </font></div>

<div class="alert alert-block alert-info">Een <b>vergelijking</b> geeft altijd een <b>bool</b> terug (True of False):<table><tr><th>betekenis</th><th>Python</th><th>betekenis</th><th>Python</th></tr><tr><td>gelijk aan</td><td><code>==</code></td><td>verschillend van</td><td><code>!=</code></td></tr><tr><td>kleiner dan</td><td><code>&lt;</code></td><td>groter dan</td><td><code>&gt;</code></td></tr><tr><td>kleiner of gelijk</td><td><code>&lt;=</code></td><td>groter of gelijk</td><td><code>&gt;=</code></td></tr></table>Met <code>in</code> en <code>not in</code> test je of iets in een lijst of string zit.</div>

### Voorbeeld 4.1 &mdash; Vergelijken

```python
ontbijt = ["boterham", "koffie", "choco"]

print(ontbijt[1] == "koffie")     # True
print(5 <= 2)                     # False
print("boter" in ontbijt)         # False (er staat "boterham")
print("koffie" not in ontbijt)    # False
```

<div class="alert alert-block alert-danger"><b>Let op &mdash; één of twee =-tekens.</b> <code>=</code> <i>kent toe</i> (een waarde in een variabele stoppen), <code>==</code> <i>vergelijkt</i> (klopt de gelijkheid?). Verwissel je ze, dan krijg je een fout of een verkeerd resultaat.</div>

<div>    <font color=#690027 markdown="1">        <h2>5. Logische operatoren</h2>     </font></div>

<div class="alert alert-block alert-info">Met de <b>logische operatoren</b> combineer je vergelijkingen:<ul><li><code>and</code>: True als <b>beide</b> waar zijn</li><li><code>or</code>: True als <b>minstens n</b> waar is</li><li><code>not</code>: draait True en False om</li></ul></div>

### Voorbeeld 5.1 &mdash; Combineren van voorwaarden

```python
temperatuur = 25
ph = 6.5

print(temperatuur > 20 and ph < 7)     # allebei waar -> True
print(temperatuur > 30 or ph < 7)      # tweede waar  -> True
print(not (ph < 7))                    # omgekeerd    -> False
```

<div class="alert alert-block alert-warning">In een labo test je vaak <b>meerdere</b> voorwaarden tegelijk: is de oplossing zuur <b>en</b> warm genoeg? Ligt een meetwaarde binnen een veilig bereik (groter dan een ondergrens <b>en</b> kleiner dan een bovengrens)? Daarvoor gebruik je <code>and</code>, <code>or</code> en <code>not</code>.</div>
