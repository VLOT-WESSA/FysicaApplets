<div>    <font color=#690027 markdown="1">        <h2>1. Keuzestructuur</h2>     </font></div>

<div class="alert alert-block alert-info">Met een <b>keuzestructuur</b> laat je de computer kiezen tussen verschillende instructies, afhankelijk van een <b>voorwaarde</b>. Je gebruikt de sleutelwoorden <b>if&nbsp;-&nbsp;elif&nbsp;-&nbsp;else</b>, wat staat voor <b>als&nbsp;-&nbsp;anders als&nbsp;-&nbsp;anders</b>.<br><br>De instructies die bij een voorwaarde horen, volgen na een dubbele punt <span style='background-color:whitesmoke; font-family:consolas;'>:</span> en worden gegroepeerd door vier spaties in te springen. Dit inspringen heet <b>indentatie</b>. Waar de inspringing stopt, stopt ook het keuzeblok.</div>

### Voorbeeld 1.1 &mdash; Aggregatietoestand van water
Voer de volgende code-cel uit. Verander daarna de waarde van `temperatuur` en voer opnieuw uit.

```python
# keuzestructuur

temperatuur = 25          # temperatuur van water in graden Celsius (bij 1 atm)

if temperatuur >= 100:
    print("Bij", temperatuur, "C is water een gas (waterdamp).")
elif temperatuur > 0:
    print("Bij", temperatuur, "C is water een vloeistof.")
else:
    print("Bij", temperatuur, "C is water vast (ijs).")

# Dit geeft: Bij 25 C is water een vloeistof.
```

<div class="alert alert-block alert-warning">Water verandert van fase bij twee vaste temperaturen (bij een druk van 1&nbsp;atm): het <b>smeltpunt</b> op 0&nbsp;&deg;C en het <b>kookpunt</b> op 100&nbsp;&deg;C. Die twee grenzen zie je rechtstreeks terug in de voorwaarden van de keuzestructuur.</div>

<div class="alert alert-block alert-danger"><b>Let op &mdash; drie veelgemaakte fouten.</b><ul><li>Gebruik <code>==</code> om te <i>vergelijken</i>, niet <code>=</code> (dat is <i>toekennen</i>).</li><li>Vergeet de dubbele punt <code>:</code> niet op het einde van de <code>if</code>-, <code>elif</code>- en <code>else</code>-regel.</li><li>Springt elk blok exact 4 spaties in? Een verkeerde indentatie geeft een <code>IndentationError</code>.</li></ul></div>

<div>    <font color=#690027 markdown="1">        <h2>2. Herhalingsstructuur</h2>     </font></div>

<div class="alert alert-block alert-info">Wil je eenzelfde instructie meerdere keren uitvoeren, of een lijst overlopen, dan gebruik je een <b>lus</b>. Er zijn twee soorten:<br>&bull; de <b>for-lus</b> voor een <i>begrensde</i> herhaling: het aantal keren ligt op voorhand vast (bv. het aantal elementen in een lijst, of een aantal dat je met <code>range()</code> vastlegt);<br>&bull; de <b>while-lus</b> voor een <i>voorwaardelijke</i> herhaling: de lus loopt door <b>zolang</b> een voorwaarde waar is.</div>

### Voorbeeld 2.1 &mdash; Een lijst meetwaarden overlopen (for-lus)
De for-lus neemt één voor één elk element uit de lijst.

```python
# for-lus over een lijst

metingen = [21.5, 22.0, 23.1, 22.8, 21.9]     # gemeten temperaturen in graden Celsius

for meting in metingen:
    print("Meting:", meting, "C")
```

### Voorbeeld 2.2 &mdash; `range()` en een valbeweging (for-lus)
Met `range(begin, einde, stap)` leg je zelf vast hoe vaak de lus draait. `range(6)` geeft de getallen 0 t.e.m. 5.

```python
# for-lus met range: vrije val

g = 9.81                              # valversnelling in m/s^2

for t in range(6):                    # t neemt de waarden 0, 1, 2, 3, 4, 5 (seconden) aan
    afstand = 0.5 * g * t**2          # s = 1/2 * g * t^2
    print("Na", t, "s is de afgelegde afstand", round(afstand, 2), "m")
```

<div class="alert alert-block alert-warning">Bij een <b>vrije val</b> (eenparig versnelde beweging) geldt $s = \tfrac{1}{2}\,g\,t^{2}$. Met een for-lus bereken je die afstand voor elke seconde in één keer, in plaats van de formule vijf keer met de hand in te vullen. Zo maak je snel een tabel van meetwaarden.</div>

<div class="alert alert-block alert-info">Bij een <b>while-lus</b> herhaal je zolang een voorwaarde waar is. Twee zaken zijn essentieel:<br>&bull; je moet de variabele in de voorwaarde <b>initialiseren</b> (een startwaarde geven) vóór de lus;<br>&bull; in de lus moet je die variabele <b>aanpassen</b>, anders blijft de voorwaarde eeuwig waar.</div>

### Voorbeeld 2.3 &mdash; Radioactief verval (while-lus)
We halveren de hoeveelheid stof telkens opnieuw en tellen hoeveel halveringstijden er nodig zijn tot er minder dan 1 gram overblijft.

```python
# while-lus: radioactief verval

hoeveelheid = 100.0                   # starthoeveelheid radioactieve stof in gram
stappen = 0                           # aantal verstreken halveringstijden

while hoeveelheid > 1.0:
    hoeveelheid = hoeveelheid / 2     # na 1 halveringstijd blijft de helft over
    stappen = stappen + 1             # verkort: stappen += 1
    print("Na", stappen, "halveringstijden:", round(hoeveelheid, 2), "g")
```

<div class="alert alert-block alert-warning">Bij <b>radioactief verval</b> halveert de hoeveelheid stof na elke <b>halveringstijd</b>. Je kan op voorhand niet vlot uitrekenen hoeveel stappen je nodig hebt om onder een drempel te geraken &mdash; net daarom is de while-lus hier ideaal: hij loopt door <i>zolang</i> er nog te veel overblijft.</div>

<div class="alert alert-block alert-danger"><b>Let op &mdash; de oneindige lus.</b> Verhoog of verlaag je de variabele in de voorwaarde niet, dan stopt de while-lus nooit. Loopt je notebook vast? Kies bovenaan <b>Kernel &rarr; Interrupt</b> (of <b>Restart</b>) om de code te onderbreken.</div>

<div>    <font color=#690027 markdown="1">        <h2>3. Geneste structuren</h2>     </font></div>

<div class="alert alert-block alert-info">Plaats je een structuur <b>binnen</b> een andere structuur, dan spreek je van een <b>geneste structuur</b>. Vaak zit er een keuzestructuur (<code>if</code>) binnen een herhalingsstructuur (<code>for</code>): voor <b>elk</b> element in een lijst neem je dan een beslissing. Elk niveau springt opnieuw 4 spaties extra in.</div>

### Voorbeeld 3.1 &mdash; pH van stalen classificeren
We overlopen een lijst gemeten pH-waarden. Voor elke waarde beslissen we of het staal zuur, neutraal of basisch is, en we tellen ondertussen het aantal zure stalen.

```python
# geneste structuur: for-lus met daarin een keuzestructuur

ph_waarden = [2.1, 7.0, 5.5, 9.3, 6.8, 12.0]   # gemeten pH van zes stalen
aantal_zuur = 0

for ph in ph_waarden:
    if ph < 7:
        soort = "zuur"
        aantal_zuur += 1
    elif ph > 7:
        soort = "basisch"
    else:
        soort = "neutraal"
    print("pH", ph, "->", soort)

print("Aantal zure stalen:", aantal_zuur)
```
