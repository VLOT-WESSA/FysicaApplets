<div>    <font color=#690027 markdown="1">        <h2>1. Lineair zoeken</h2>     </font></div>

<div class="alert alert-block alert-info">Bij <b>lineair zoeken</b> overloop je de lijst element per element tot je de gezochte waarde vindt. Je geeft de <b>index</b> terug (de plaats), of $-1$ als de waarde niet voorkomt. Dit werkt op elke lijst, ook een ongesorteerde.</div>

### Voorbeeld 1.1 &mdash; Een waarde zoeken

```python
def lineair_zoeken(lijst, doel):
    for i in range(len(lijst)):
        if lijst[i] == doel:
            return i          # gevonden op index i
    return -1                 # niet gevonden

metingen = [23, 41, 15, 8, 42, 16]
print(lineair_zoeken(metingen, 42))   # 4
print(lineair_zoeken(metingen, 99))   # -1
```

<div>    <font color=#690027 markdown="1">        <h2>2. Binair zoeken</h2>     </font></div>

<div class="alert alert-block alert-info">In een <b>gesorteerde</b> lijst kan je veel sneller zoeken met <b>binair zoeken</b>. Je kijkt naar het middelste element: is het te groot, dan zoek je verder in de linkerhelft; is het te klein, in de rechterhelft. Zo halveer je elke stap de zoekruimte.</div>

### Voorbeeld 2.1 &mdash; Binair zoeken

```python
def binair_zoeken(lijst, doel):
    links = 0
    rechts = len(lijst) - 1
    while links <= rechts:
        midden = (links + rechts) // 2
        if lijst[midden] == doel:
            return midden
        elif lijst[midden] < doel:
            links = midden + 1     # zoek rechts verder
        else:
            rechts = midden - 1    # zoek links verder
    return -1

gesorteerd = [3, 8, 15, 16, 23, 41, 42]
print(binair_zoeken(gesorteerd, 23))   # 4
print(binair_zoeken(gesorteerd, 10))   # -1
```

<div class="alert alert-block alert-warning">In een lijst van 1000 elementen vindt lineair zoeken de waarde na maximaal 1000 stappen; binair zoeken lukt in hooguit 10 stappen (want $2^{10} > 1000$). Voor grote datasets is de keuze van het algoritme dus cruciaal.</div>

<div class="alert alert-block alert-danger"><b>Let op.</b> Binair zoeken werkt <b>alleen</b> op een <b>gesorteerde</b> lijst. Op een ongeordende lijst geeft het foute resultaten. Sorteer dus eerst, of gebruik lineair zoeken.</div>

<div>    <font color=#690027 markdown="1">        <h2>3. Sorteren: selectiesort</h2>     </font></div>

<div class="alert alert-block alert-info">Bij <b>selectiesort</b> zoek je telkens het kleinste element uit het nog niet gesorteerde deel en zet je het vooraan. Je herhaalt dit tot de hele lijst geordend is. Zo ontwerp je zelf een sorteeralgoritme.</div>

### Voorbeeld 3.1 &mdash; Selectiesort

```python
def selectiesort(lijst):
    lijst = list(lijst)               # werk op een kopie
    for i in range(len(lijst)):
        kleinste = i
        for j in range(i + 1, len(lijst)):
            if lijst[j] < lijst[kleinste]:
                kleinste = j
        lijst[i], lijst[kleinste] = lijst[kleinste], lijst[i]   # wisselen
    return lijst

print(selectiesort([23, 41, 15, 8, 42, 16]))   # [8, 15, 16, 23, 41, 42]
```

<div class="alert alert-block alert-info">Python heeft ook een ingebouwde sortering: <code>sorted(lijst)</code> geeft een nieuwe gesorteerde lijst, en <code>lijst.sort()</code> sorteert ter plaatse. In de praktijk gebruik je die; zelf sorteren leert je vooral <i>hoe</i> het werkt.</div>

<div>    <font color=#690027 markdown="1">        <h2>4. Algoritmisch denken op meetgegevens</h2>     </font></div>

<div class="alert alert-block alert-info">Met dezelfde denkwijze bereken je zelf grootheden op een lijst, zonder kant-en-klare functies: het maximum zoeken, tellen onder een voorwaarde, of een positie bepalen.</div>

### Voorbeeld 4.1 &mdash; Zelf het maximum en zijn plaats zoeken

```python
metingen = [23, 41, 15, 8, 42, 16]

grootste = metingen[0]
plaats = 0
for i in range(len(metingen)):
    if metingen[i] > grootste:
        grootste = metingen[i]
        plaats = i

print("grootste waarde:", grootste, "op index", plaats)
```

<div>    <font color=#690027 markdown="1">        <h2>5. Efficiëntie: doorlooptijd en geheugengebruik</h2>     </font></div>

<div class="alert alert-block alert-info">Voor eenzelfde probleem bestaan vaak <b>meerdere</b> algoritmen. Om te kiezen welk het <b>beste</b> is, kijk je naar twee dingen:<br>&bull; de <b>doorlooptijd</b>: hoe lang duurt de berekening?<br>&bull; het <b>geheugengebruik</b>: hoeveel extra geheugen is nodig?<br>Je meet de doorlooptijd met de module <code>time</code>: neem de tijd vóór en ná de berekening en trek af.</div>

### Voorbeeld 5.1 &mdash; Twee sorteermethodes vergelijken
We sorteren dezelfde grote lijst met onze eigen selectiesort en met de ingebouwde `sorted()`, en meten telkens de tijd.

```python
import time
import random

data = [random.randint(0, 10000) for i in range(2000)]   # 2000 willekeurige getallen

start = time.time()
selectiesort(data)                 # onze eigen methode
print("selectiesort:", round(time.time() - start, 4), "s")

start = time.time()
sorted(data)                       # ingebouwde methode
print("sorted():    ", round(time.time() - start, 4), "s")
```

<div class="alert alert-block alert-warning">Onze <b>selectiesort</b> doorloopt de lijst voor elk element opnieuw: bij 2000 getallen zijn dat al miljoenen vergelijkingen, terwijl <code>sorted()</code> een slimmer algoritme gebruikt en veel sneller klaar is. Voor grote datasets is de keuze van het algoritme dus doorslaggevend voor de doorlooptijd.</div>

<div class="alert alert-block alert-info">Ook het <b>geheugengebruik</b> verschilt. Sorteer je <i>ter plaatse</i> met <code>lijst.sort()</code>, dan gebruik je nauwelijks extra geheugen. Maak je een <b>kopie</b> (zoals <code>sorted()</code> of onze selectiesort met <code>list(lijst)</code>), dan staat de lijst even <b>dubbel</b> in het geheugen. Bij grote datasets kan dat verschil belangrijk zijn. Een <b>NumPy-array</b> gebruikt bovendien minder geheugen dan een gewone lijst met dezelfde getallen.</div>

<div class="alert alert-block alert-danger"><b>Analyse maakt het verschil.</b> Kies je algoritme bewust: voor een <i>kleine</i> lijst maakt het niet uit, maar voor <i>grote</i> datasets bepaalt de methode of je resultaat er in een oogwenk of pas na minuten is &mdash; en of het al dan niet in het geheugen past.</div>
