### De nodige modules importeren

```python
import numpy as np
import matplotlib.pyplot as plt
```

<div>    <font color=#690027 markdown="1">        <h2>1. Een puntenwolk tekenen</h2>     </font></div>

<div class="alert alert-block alert-info">Met <code>plt.scatter(x, y)</code> teken je de koppels als losse punten. Je geeft de x-coördinaten en de y-coördinaten elk als een NumPy-array.</div>

### Voorbeeld 1.1 &mdash; Vijf meetpunten plotten

```python
x = np.array([0, 1, 2, 3, 4])
y = np.array([3, 24, 12, 48, 6])

plt.figure(figsize=(6, 4))
plt.scatter(x, y)
plt.title("Spreidingsdiagram")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

<div>    <font color=#690027 markdown="1">        <h2>2. Data uit een CSV-bestand</h2>     </font></div>

<div class="alert alert-block alert-info">Meetgegevens staan vaak in een <b>CSV-bestand</b> (comma separated values). Je leest het in met de module <b>pandas</b> tot een <b>DataFrame</b> (een tabel). Met <code>head()</code> en <code>tail()</code> bekijk je het begin en het einde. Een kolom zet je om naar een NumPy-array met <code>.values</code>.</div>

### Voorbeeld 2.1 &mdash; Een CSV inlezen en omzetten
*Deze cel vereist een bestand `punten.csv` met kolommen `x` en `y` in dezelfde map.*

```python
import pandas as pd

tabel = pd.read_csv("punten.csv")     # inlezen
print(tabel.head())                   # eerste rijen

x = tabel["x"].values                 # kolom -> NumPy-array
y = tabel["y"].values
print(x)
print(y)
```

<div class="alert alert-block alert-warning">CSV is het standaardformaat waarin meettoestellen en dataloggers hun gegevens bewaren. Met pandas haal je in twee regels een hele meetreeks binnen en zet je ze klaar om te plotten of te analyseren.</div>

<div>    <font color=#690027 markdown="1">        <h2>3. Het bereik van de assen</h2>     </font></div>

<div class="alert alert-block alert-info">Om je grafiek leesbaar te maken, stel je het <b>bereik</b> van de assen in met <code>plt.axis([xmin, xmax, ymin, ymax])</code>. De grenzen bepaal je uit de data met <code>np.min</code> en <code>np.max</code>.</div>

### Voorbeeld 3.1 &mdash; Passend bereik kiezen

```python
x = np.array([0, 1, 2, 3, 4])
y = np.array([3, 24, 12, 48, 6])

plt.figure(figsize=(6, 4))
plt.scatter(x, y)
plt.axis([np.min(x) - 1, np.max(x) + 1, 0, np.max(y) + 5])   # wat marge rond de data
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

<div>    <font color=#690027 markdown="1">        <h2>4. Samenhang meten: de correlatiecoëfficiënt</h2>     </font></div>

<div class="alert alert-block alert-info">De <b>correlatiecoëfficiënt</b> $r$ drukt uit hoe sterk twee grootheden lineair samenhangen. $r$ ligt tussen $-1$ en $+1$: dicht bij $+1$ betekent een sterk stijgend verband, dicht bij $-1$ een sterk dalend verband, en dicht bij $0$ nauwelijks lineair verband. Je berekent $r$ met <code>np.corrcoef(x, y)[0, 1]</code>.</div>

### Voorbeeld 4.1 &mdash; Correlatie berekenen

```python
x = np.array([0, 1, 2, 3, 4, 5])
y = np.array([1.9, 4.1, 5.8, 8.2, 9.9, 12.1])   # bijna een rechte

r = np.corrcoef(x, y)[0, 1]
print("correlatiecoefficient r =", round(r, 3))
```

<div class="alert alert-block alert-danger"><b>Let op &mdash; correlatie is niet hetzelfde als verband.</b> Twee heel verschillende puntenwolken kunnen dezelfde $r$ hebben. Bekijk daarom altijd óók het spreidingsdiagram; vertrouw nooit op $r$ alleen.</div>

### Voorbeeld 4.2 &mdash; Dezelfde correlatie, andere vormTwee datasets met dezelfde x maar een andere y kunnen bijna hetzelfde $r$ geven (naar het kwartet van Anscombe).

```python
x  = np.array([10, 8, 13, 9, 11, 14, 6, 4, 12, 7, 5])
y1 = np.array([8.04, 6.95, 7.58, 8.81, 8.33, 9.96, 7.24, 4.26, 10.84, 4.82, 5.68])
y2 = np.array([9.14, 8.14, 8.74, 8.77, 9.26, 8.10, 6.13, 3.10, 9.13, 7.26, 4.74])

print("r voor dataset 1:", round(np.corrcoef(x, y1)[0, 1], 3))
print("r voor dataset 2:", round(np.corrcoef(x, y2)[0, 1], 3))

plt.figure(figsize=(8, 3))
plt.subplot(1, 2, 1); plt.scatter(x, y1); plt.title("dataset 1")
plt.subplot(1, 2, 2); plt.scatter(x, y2); plt.title("dataset 2")
plt.show()
```
