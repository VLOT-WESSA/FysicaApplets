### Voorbeeld 2.1 &mdash; Massadefect en bindingsenergie
Bij het vormen van een kern is de massa kleiner dan de som van de losse deeltjes. Dat massaverschil (het *massadefect*) komt vrij als energie: $E = \Delta m \cdot c^{2}$.

```python
massa_p = 1.007276      # massa proton (u)
massa_n = 1.008665      # massa neutron (u)
massa_He = 4.002602     # massa He-4 kern (u)

massadefect = 2 * massa_p + 2 * massa_n - massa_He   # He-4 = 2 p + 2 n
energie_MeV = massadefect * 931.5                    # 1 u komt overeen met 931,5 MeV
print("massadefect:", round(massadefect, 5), "u")
print("bindingsenergie:", round(energie_MeV, 2), "MeV")
```

<div class="alert alert-block alert-warning">Het <b>massadefect</b> vertaalt zich via $E=\Delta m\,c^{2}$ in de <b>bindingsenergie</b> die een kern bijeenhoudt. De omrekening 1&nbsp;u $\rightarrow$ 931,5&nbsp;MeV gebruik je voortdurend in de kernfysica.</div>

### Voorbeeld 3.1 &mdash; Stabiel of niet?
Voor lichte kernen zijn stabiele kernen ongeveer die waarvoor het aantal neutronen N ongeveer gelijk is aan het aantal protonen Z.

```python
Z = 6      # aantal protonen (koolstof)
N = 8      # aantal neutronen (C-14)

if N == Z:
    print("N = Z: vaak stabiel")
elif N > Z:
    print("neutronenoverschot: mogelijk beta-min-verval")
else:
    print("protonenoverschot: mogelijk beta-plus-verval")
```

### Voorbeeld 4.1 &mdash; Verval over gelijke tijdstappen
Na elke halveringstijd blijft de helft over. We volgen 1000 kernen over 5 halveringstijden.

```python
N = 1000        # beginaantal kernen
for stap in range(1, 6):
    N = N / 2
    print("na", stap, "halveringstijden:", round(N, 1), "kernen")
```

<div>    <font color=#690027 markdown="1">        <h2>5. Functies</h2>     </font></div>

<div class="alert alert-block alert-info">Een functie definieer je met <code>def naam(parameters):</code> en geeft een resultaat terug met <code>return</code>. Zo bewaar je een formule en hergebruik je ze.</div>

### Voorbeeld 5.1 &mdash; De vervalwet als functie
Het aantal kernen na tijd $t$ is $N(t) = N_0 \cdot \left(\tfrac12\right)^{t/T}$, met $T$ de halveringstijd.

```python
def verval(N0, halveringstijd, t):
    return N0 * 0.5 ** (t / halveringstijd)

print(round(verval(1000, 5730, 5730), 1))    # na 1 halveringstijd -> 500.0
print(round(verval(1000, 5730, 11460), 1))   # na 2 halveringstijden -> 250.0
```

<div class="alert alert-block alert-warning">C-14 heeft een halveringstijd van 5730 jaar. Door de resterende hoeveelheid C-14 in een vondst te meten en met deze vervalwet te vergelijken, bepalen archeologen de ouderdom (<b>koolstofdatering</b>).</div>
