<div>    <font color=#690027 markdown="1">        <h1>DATATYPES</h1>     </font></div>

<div class="alert alert-box alert-success">Elk stukje informatie in Python heeft een <b>type</b>: een getal, tekst, een lijst... In dit hoofdstuk leer je de belangrijkste datatypes kennen en hoe je ze omvormt.<br><br><b>Wat kan je na dit hoofdstuk?</b><ul><li>de basistypes <b>str, int, float, bool</b> en <b>tuple</b> herkennen en gebruiken;</li><li>het type van een object opvragen en omvormen (<b>typecasting</b>);</li><li>werken met de collecties <b>list</b> en <b>dictionary</b>;</li><li>kennismaken met de <b>NumPy-array</b> voor wetenschappelijke gegevens.</li></ul></div>

<b>Leeswijzer van de kaders</b><br>&bull; <span style='color:#31708f'><b>Blauw</b></span> = theorie &nbsp;&bull; <span style='color:#8a6d3b'><b>Geel</b></span> = achtergrond en toepassingen &nbsp;&bull; <span style='color:#a94442'><b>Rood</b></span> = let op / veelgemaakte fout<br>De opbouw is telkens: eerst de <b>theorie</b>, dan een uitgewerkt <b>voorbeeld</b>, dan een <b>oefening</b> voor jou.

<div class="alert alert-block alert-info">Alles in Python is een <b>object</b>. Elk object heeft een <b>type</b>. Je vraagt het type op met de functie <code>type()</code>. Een object maak je door er met <code>=</code> een waarde aan toe te kennen; de naam links van het <code>=</code>-teken heet een <b>variabele</b>.</div>

### Voorbeeld 0.1 &mdash; Het type opvragen
Voer de cellen uit en let op wat `type()` teruggeeft.

```python
bloem = "iris"      # tekst tussen aanhalingstekens
print(type(bloem))

getal = -3          # een geheel getal
print(type(getal))
```

<div class="alert alert-block alert-info">Een <b>variabele</b> is een naam waaraan je een waarde koppelt; die waarde kan tijdens het programma <b>veranderen</b>. Een <b>constante</b> is een waarde die <b>niet</b> verandert, zoals de lichtsnelheid of de valversnelling. Python kent geen echt aparte constante, maar er is een <b>afspraak</b>: je schrijft de naam van een constante volledig in <b>HOOFDLETTERS</b>. Zo zie je meteen dat je die waarde niet mag wijzigen.</div>

### Voorbeeld 0.2 &mdash; Variabele en constante

```python
LICHTSNELHEID = 3.0e8      # constante (verandert niet) -> HOOFDLETTERS
G = 9.81                   # constante: valversnelling

hoogte = 10                # variabele (mag veranderen)
hoogte = hoogte + 5        # de variabele wijzigt

print(LICHTSNELHEID, G, hoogte)
```

<div class="alert alert-block alert-danger"><b>Let op.</b> Python <b>verhindert</b> niet dat je een constante toch overschrijft &mdash; de HOOFDLETTERS zijn enkel een afspraak tussen programmeurs. Behandel een waarde in HOOFDLETTERS dus als vast en wijzig ze niet.</div>

<div>    <font color=#690027 markdown="1">        <h2>1. De basistypes</h2>     </font></div>

<div class="alert alert-block alert-info"><b>str (string)</b> is het type van <b>tekst</b>: een opeenvolging van karakters tussen aanhalingstekens (<code>"..."</code> of <code>'...'</code>). Elk karakter heeft een <b>index</b>; je begint te tellen vanaf <b>0</b>. Met <code>len()</code> vraag je het aantal karakters op. Strings samenvoegen met <code>+</code> heet <b>concateneren</b>.</div>

### Voorbeeld 1.1 &mdash; Werken met een string

```python
bloem = "iris"

print(bloem[0])          # eerste karakter -> i
print(bloem[-1])         # laatste karakter -> s
print(len(bloem))        # aantal karakters -> 4

kleur = "gele "
print(kleur + bloem)     # concateneren -> gele iris
```

<div class="alert alert-block alert-info"><b>int (integer)</b> is een <b>geheel getal</b>, positief of negatief, zonder decimalen.<br><b>float</b> is een <b>kommagetal</b>. Let op: je gebruikt een <b>punt</b>, geen komma. Je kan ook de wetenschappelijke notatie gebruiken: <code>4.75e3</code> betekent $4{,}75\cdot 10^{3}$.</div>

### Voorbeeld 1.2 &mdash; Gehele en kommagetallen

```python
massa = 42               # int
print(type(massa))

versnelling = 9.81       # float
print(type(versnelling))

lichtsnelheid = 3.0e8    # 3,0 x 10^8  (wetenschappelijke notatie)
print(lichtsnelheid)
```

<div class="alert alert-block alert-danger"><b>Let op &mdash; de komma.</b> Schrijf je een kommagetal als <code>4,75</code> (met een komma), dan denkt Python dat je <b>twee</b> waarden bedoelt. Gebruik altijd een <b>punt</b>: <code>4.75</code>.</div>

<div class="alert alert-block alert-info"><b>bool (boolean)</b> heeft slechts twee mogelijke waarden: <b>True</b> of <b>False</b>. Zo'n waarde is het resultaat van een vergelijking. Een bool is eigenlijk een speciaal geheel getal: <code>True</code> telt als 1, <code>False</code> als 0.</div>

### Voorbeeld 1.3 &mdash; Een booleaanse waarde

```python
print(2 < 5)             # True
print(10 == 3)           # False
print(type(2 < 5))       # bool
```

<div class="alert alert-block alert-info"><b>tuple</b> is een opeenvolging van elementen die bij elkaar horen en waarvan de volgorde vastligt. Je noteert de elementen tussen <b>ronde haakjes</b>. Een tuple kan je niet meer wijzigen nadat ze gemaakt is.</div>

### Voorbeeld 1.4 &mdash; Een cordinaat als tuple

```python
punt = (5.0, 3.5)        # een (x, y)-cordinaat
print(punt[0])           # x-cordinaat -> 5.0
print(punt[1])           # y-cordinaat -> 3.5
```

<div>    <font color=#690027 markdown="1">        <h2>2. Typecasting</h2>     </font></div>

<div class="alert alert-block alert-info">Je kan het type van een object <b>omvormen</b> naar een ander type. Dat heet <b>typecasting</b>. Je gebruikt <code>int()</code>, <code>float()</code> of <code>str()</code>.<br>Dit is nodig omdat je bv. een getal en tekst niet zomaar kan samenvoegen: zet het getal eerst om naar een string.</div>

### Voorbeeld 2.1 &mdash; Omvormen tussen types

```python
aantal = 5.0
aantal_geheel = int(aantal)          # float -> int  =>  5
print(aantal_geheel)

zin = str(aantal_geheel) + " bloemen"  # int -> str om te concateneren
print(zin)                             # 5 bloemen
```

<div class="alert alert-block alert-warning">Bij het inlezen van meetgegevens is typecasting onmisbaar. Getallen die je uit een bestand of van de gebruiker krijgt, zijn eerst tekst (<b>str</b>). Voor je ermee rekent, zet je ze om naar <b>int</b> of <b>float</b>.</div>

<div>    <font color=#690027 markdown="1">        <h2>3. Collectie: list</h2>     </font></div>

<div class="alert alert-block alert-info">Een <b>list (lijst)</b> is een geordende rij elementen tussen <b>vierkante haakjes</b> <code>[ ]</code>. Een lijst mag meerdere datatypes bevatten en eenzelfde element mag meermaals voorkomen. Elk element heeft een <b>index</b> (starten vanaf 0). Met <code>len()</code> vraag je het aantal elementen op.</div>

### Voorbeeld 3.1 &mdash; Een lijst en zijn indicesBloemplanten worden opgedeeld in monocotylen en dicotylen. We bewaren enkele voorbeelden in lijsten.

```python
monocotyl = ["prei", "orchidee", "iris", "ananasboom"]
dicotyl   = ["roos", "tomatenplant", "eik", "zonnebloem"]

print(len(monocotyl))     # 4
print(dicotyl[0])         # eerste element -> roos
print(dicotyl[-1])        # laatste element -> zonnebloem
```

<div class="alert alert-block alert-info"><b>Slicing</b> is het selecteren van een deel van een lijst: <code>lijst[start:stop:stap]</code>. Het element met index <code>stop</code> hoort er <b>niet</b> meer bij (halfopen interval). Concateneren met <code>+</code> werkt ook bij lijsten. Strings kan je op dezelfde manier slicen.</div>

### Voorbeeld 3.2 &mdash; Slicing en samenvoegen

```python
dicotyl = ["roos", "tomatenplant", "eik", "zonnebloem"]

print(dicotyl[1:3])       # index 1 t.e.m. 2 -> ['tomatenplant', 'eik']
print(dicotyl[:2])        # begin t.e.m. index 1
print(dicotyl[::2])       # elk tweede element

bloemplanten = ["prei", "iris"] + dicotyl   # lijsten samenvoegen
print(bloemplanten)
```

<div class="alert alert-block alert-info">Op een lijst kan je <b>methodes</b> toepassen. Je roept ze op met een punt: <code>lijst.methode()</code>.<ul><li><code>append(element)</code>: voegt achteraan toe</li><li><code>insert(index, element)</code>: voegt toe op een plaats</li><li><code>pop(index)</code>: verwijdert op een plaats</li><li><code>remove(element)</code>: verwijdert de eerste met die waarde</li><li><code>sort()</code>: sorteert de lijst</li></ul></div>

### Voorbeeld 3.3 &mdash; Een lijst wijzigen en sorteren

```python
getallen = [3, 7, 1, 13, 2]
getallen.append(5)        # -> [3, 7, 1, 13, 2, 5]
getallen.sort()           # -> [1, 2, 3, 5, 7, 13]
print(getallen)

wetenschappers = ["Darwin", "Newton", "Einstein", "Curie"]
wetenschappers.sort()     # alfabetisch
print(wetenschappers)
```

<div>    <font color=#690027 markdown="1">        <h2>4. Collectie: dictionary</h2>     </font></div>

<div class="alert alert-block alert-info">Een <b>dictionary (woordenboek)</b> bestaat uit paren van een <b>sleutel (key)</b> en een bijhorende <b>waarde (value)</b>, tussen <b>accolades</b> <code>{ }</code>, gescheiden door een dubbele punt. Je zoekt een waarde op via haar sleutel, niet via een index. De volgorde van de paren speelt geen rol.</div>

### Voorbeeld 4.1 &mdash; Fotosynthesetype per plant
Sommige planten gebruiken C3-fotosynthese, andere C4. We bewaren dat in een dictionary.

```python
planten = {
    "tarwe": "C3",
    "mais": "C4",
    "suikerriet": "C4",
    "rijst": "C3"
}

print(planten["mais"])          # waarde bij sleutel "mais" -> C4
print(len(planten))             # aantal paren -> 4

planten["gerst"] = "C3"         # een paar toevoegen
print("rijst" in planten)       # zit de sleutel erin? -> True
```

<div class="alert alert-block alert-info">Nuttige methodes op een dictionary: <code>get(key)</code> (waarde ophalen), <code>keys()</code> (alle sleutels), <code>values()</code> (alle waarden), <code>items()</code> (alle paren), <code>pop(key)</code> (paar verwijderen).</div>

### Voorbeeld 4.2 &mdash; Sleutels en waarden

```python
massa_g_per_mol = {"H": 1, "C": 12, "O": 16, "N": 14}   # atoommassa's

print(massa_g_per_mol.get("C"))     # 12
print(massa_g_per_mol.keys())       # alle symbolen
print(massa_g_per_mol.values())     # alle massa's
```

<div class="alert alert-block alert-warning">Een dictionary is ideaal om wetenschappelijke gegevens te koppelen: een chemisch symbool aan zijn atoommassa, een plantnaam aan zijn fotosynthesetype, een land aan zijn hoofdstad. Je zoekt dan bliksemsnel op via de sleutel.</div>

<div>    <font color=#690027 markdown="1">        <h2>5. Wetenschappelijke gegevens: de NumPy-array</h2>     </font></div>

<div class="alert alert-block alert-info">Een <b>NumPy-array</b> is een tabel van getallen van <b>hetzelfde type</b>. Ze lijkt op een lijst, maar je kan er rechtstreeks op rekenen en ze is sneller voor grote hoeveelheden data. Je hebt de <b>module numpy</b> nodig, die je invoert met <code>import numpy as np</code>.</div>

### Voorbeeld 5.1 &mdash; Rekenen op een hele array

```python
import numpy as np

metingen = np.array([28, 40, 41, 44, 44, 45, 51])   # meetreeks

print(metingen.mean())     # gemiddelde
print(metingen.max())      # grootste waarde
print(metingen * 2)        # elk element maal 2, in n keer
```

<div class="alert alert-block alert-warning">Bij een echte meetreeks (bv. temperaturen, massa's, spanningen) reken je met NumPy in één regel het gemiddelde, het minimum of het maximum uit. Dat is precies wat je in een labo nodig hebt om metingen te verwerken.</div>

<div class="alert alert-block alert-danger"><b>Let op.</b> Een NumPy-array vraagt de module <code>numpy</code>. Vergeet je de regel <code>import numpy as np</code> bovenaan, dan krijg je een <code>NameError</code>.</div>
